# Theory CMB lensing auto-spectrum $C_{\ell}^{\kappa \kappa}$

## setting up CAMB

```python
def load_cosmology(filename):
    with open(filename) as f:
        return json.load(f)
```

```python
def get_camb_results(
    cosmo_dict,
    z_pk,
    kmax=100.0,
    lmax=2500,
    nonlinear=False,
):
    """
    Read a cosmological parameter dictionary, extract cosmological parameters,
    and build CAMB results instance
    """

    # Convert to CAMB inputs

    H0 = 100.0 * cosmo_dict["h"]
    ombh2 = cosmo_dict["omega_B"] * cosmo_dict["h"]**2
    omch2 = (cosmo_dict["omega_M"] - cosmo_dict["omega_B"]) * cosmo_dict["h"]**2

    # Initial CAMB params

    pars = camb.CAMBparams()
    pars.set_cosmology(
        H0=H0,
        ombh2=ombh2,
        omch2=omch2,
    )
    pars.set_dark_energy(w=cosmo_dict["w"])
    pars.InitPower.set_params(
        ns=cosmo_dict["ns"],
        As=cosmo_dict["As"]
    )

    # Matter power settings

    pars.set_matter_power(
        redshifts=z_pk,
        kmax=kmax
    )

    # note that to get CAMB's internal calculation accurate at the 1% level at L~2000,
    # need lens_potential_accuracy=2. Increase to 4 for accurate match to the Limber calculation here
    if nonlinear:
        pars.NonLinear = camb.model.NonLinear_pk
        pars.set_for_lmax(lmax=lmax, lens_potential_accuracy=4, nonlinear=True)
    else:
        pars.set_for_lmax(lmax=lmax, lens_potential_accuracy=4, nonlinear=False)

    # Final results
    results = camb.get_results(pars)

    return results
```

## manual Limber integral

```python
def W_kappa(z, chi, chi_source, cosmo_params):
    '''
    CMB lensing convergence kernel
    '''
    
    # scale factor at redshift of lensing structure
    a = 1.0 / (1.0 + z)

    # cosmological parameters
    H0, om = cosmo_params
    c_kms = c * 1E-3
    
    # prefactor of the kernel
    W_prefactor = (1.5) * (H0/c_kms)**2 * om * (chi_source - chi)/chi_source
    W_geom = chi / a
    
    return W_prefactor * W_geom
```

```python
def limber(z, 
           ell, 
           kmax, 
           chi_source, 
           cosmo_params, 
           pk_interp, 
           results,
           kp=0.312, # default fitted smoothing scale in (1/Mpc)
):
    '''
    CMB lensing auto-spectrum using Limber approximation
    '''

    # chi(z) and wavenumber of Limber approximation and 
    chi = results.comoving_radial_distance(z)
    k = (ell + 0.5) / chi

    # jacobian dchi/dz of the transformation integrating over z instead of chi
    c_kms = c * 1e-3
    Hz = results.hubble_parameter(z)

    # avoid extrapolation of CAMB kmax and small scales
    if 1e-4 < k < kmax:

        # tracer kernel
        W1 = W_kappa(z, chi, chi_source, cosmo_params)
        W2 = W1
        # matter power spectrum
        pk = pk_interp.P(z, k, grid=False)
        # smoothing
        smooth = np.exp(-(k / kp)**2)

        return (c_kms / Hz) * ((W1 * W2) / chi**2) * pk * smooth

    else:
        return 0
```

```python
def get_limber_cls_kk(
        cosmo_params_filename, 
        z_pk,
        kmax=100, 
        lmin=2,
        lmax=2500
):

    '''
    Compute the angular kk power spectra for fitting using the Limber approximation.
    '''
    
    # get linear and non-linear matter power spectrum
    cosmo_dict = load_cosmology(cosmo_params_filename)
    z_max, z_out = cosmo_dict["z_max"], cosmo_dict["z_out"]

    results_lin = get_camb_results(
        cosmo_dict=cosmo_dict,
        z_pk=z_pk,
        kmax=kmax,
        lmax=lmax,
        nonlinear=False)
    results_nlin = get_camb_results(
        cosmo_dict=cosmo_dict,
        z_pk=z_pk,
        kmax=kmax,
        lmax=lmax,
        nonlinear=True)
    
    # comoving distance of the last scattering surface 
    z_star = results_lin.get_derived_params()['zstar']  # z_star doesn't match z_out until third box
    chi_star = results_lin.comoving_radial_distance(z_star)
    chi_max = results_lin.comoving_radial_distance(z_max)

    # Hubble parameter and matter density for the kernel
    H0 = cosmo_dict["h"] * 100
    om = cosmo_dict["omega_M"]
    cosmo_params = (H0, om)

    # get the matter power spectrum interpolators for linear and non-linear cases
    pk_interp_lin = results_lin.get_matter_power_interpolator(nonlinear=False, 
                                                      hubble_units=False, 
                                                      k_hunit=False,  
                                                      var1='delta_tot', 
                                                      var2='delta_tot')
    pk_interp_nlin = results_nlin.get_matter_power_interpolator(nonlinear=True, 
                                                         hubble_units=False, 
                                                         k_hunit=False,  
                                                         var1='delta_tot', 
                                                         var2='delta_tot')
    
    #store the cls from linear and nonlinear limber approximation integration
    # redshift integration up to zmax
    ells = np.arange(lmin, lmax+1)
    cl_kk_lin_zs_zs = np.zeros(ells.shape)
    cl_kk_nlin_zs_zs = np.zeros(ells.shape)
    cl_kk_lin_zmax_zs = np.zeros(ells.shape)
    cl_kk_nlin_zmax_zs = np.zeros(ells.shape)
    #cl_kk_lin_zmax_zmax = np.zeros(ells.shape)
    #cl_kk_nlin_zmax_zmax = np.zeros(ells.shape)

    # manually integrate the limber approximation
    for i, ell in enumerate(ells):

        # integral up to redshift z_source, source plane at z_star
        chi_source = chi_star
        z_cut = z_star
        cl_kk_lin_zs_zs[i] = quad(limber, 1e-4, z_cut, args=(ell, kmax, chi_source, cosmo_params, pk_interp_lin, results_lin), epsabs=1E-10)[0]
        cl_kk_nlin_zs_zs[i] = quad(limber, 1e-4, z_cut, args=(ell, kmax, chi_source, cosmo_params, pk_interp_nlin, results_nlin), epsabs=1E-10)[0]
        
        # integral up to redshift z_max, source plane at z_star
        chi_source = chi_star
        z_cut = z_max
        cl_kk_lin_zmax_zs[i] = quad(limber, 1e-4, z_cut, args=(ell, kmax, chi_source, cosmo_params, pk_interp_lin, results_lin), epsabs=1E-10)[0]
        cl_kk_nlin_zmax_zs[i] = quad(limber, 1e-4, z_cut, args=(ell, kmax, chi_source, cosmo_params, pk_interp_nlin, results_nlin), epsabs=1E-10)[0]

        # integral up to redshift z_max, source plane at z_max, uncomment just for the first box
        #chi_source = chi_max
        #z_cut = z_max
        #cl_kk_lin_zmax_zmax[i] = quad(limber, 1e-4, z_max, args=(ell, kmax, chi_source, cosmo_params, pk_interp_lin, results_lin), epsabs=1E-10)[0]
        #cl_kk_nlin_zmax_zmax[i] = quad(limber, 1e-4, z_max, args=(ell, kmax, chi_source, cosmo_params, pk_interp_nlin, results_nlin), epsabs=1E-10)[0]

    return {
        'ells': ells,
        'results_lin': results_lin,
        'results_nlin': results_nlin,
        'cl_kk_lin_zmax_zs': cl_kk_lin_zmax_zs,
        'cl_kk_nlin_zmax_zs': cl_kk_nlin_zmax_zs,
        'cl_kk_lin_zs_zs': cl_kk_lin_zs_zs,
        'cl_kk_nlin_zs_zs': cl_kk_nlin_zs_zs,
        'z_max': float(z_max),
        'z_out': float(z_out),
        'z_star': z_star,
    }
```

```python
# set the redshift range for the boltzmann solver
zmin, zmax = 0, 1100
nz = 200
z_pk = zmin + np.exp(np.log(zmax - zmin + 1)*np.linspace(0, 1, nz)) - 1
z_pk = z_pk[::-1] # reverse to have higher redshift first for better interpolation at high z
lmin, lmax = 2, 2500
```

## full sky (non-Limber) integral

```python
def get_camb_cls(results, lmin=2, lmax=2500):
    '''
    Get the CAMB kappa power spectrum Cl^kappakappa from the CAMB results object
    '''
    camb_cls = results.get_lens_potential_cls(lmax=lmax, raw_cl=True)
    camb_cls = camb_cls[lmin:, 0]
    ells = np.arange(lmin, lmax+1)
    camb_cls = (ells * (ells + 1) / 2 )**2 * camb_cls  # Convert from C_ell^phiphi to C_ell^kappakappa

    return camb_cls # Extract the lensing potential auto-spectrum (phiphi)
```

```python
def plot_cls_comparison(ells_colore, cl_kk_colore, err_kk, ells, cl_kk_lin, cl_kk_lin_rebin, z_out,z_star, box_label):

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(6, 6), height_ratios=[2.5, 1], dpi=120, sharex=True)

    # compute the reduced chi-squared between the rebinned linear limber and the rebinned CoLoRe cls
    chi2 = np.sum((cl_kk_colore - cl_kk_lin_rebin)**2 / err_kk**2)
    chi2_nu = chi2 / (len(ells_colore))

    # plot the rebinned CoLoRe Cls with error bars
    ax1.errorbar(ells_colore, cl_kk_colore, yerr=err_kk, elinewidth=0.7, capsize=1.0, fmt='C0.', markersize=5.0, 
                 label=rf'CoLoRe_{box_label} $(z_{{\rm max}} = {z_max}, \ z_s={int(z_out)})$')
    ax1.scatter(0, 0 , s=0, label=rf'$\chi^2_\nu = {chi2_nu:.2f}$')

    # manual limber integration full redshift
    ax1.plot(ells, cl_kk_lin, c='C0', ls='-', label=rf'Limber_linear ($z_{{\rm cut}} = {z_max}, \ z_s={int(z_star)}$)')

    ax1.set_xscale('log')
    ax1.set_yscale('log')
    ax1.set_xlim(1.5, lmax)
    ax1.set_ylabel(r'$C_{L}^{\kappa \kappa}$')
    ax1.legend()

    # plot the relative error between the rebinned linear limber and the rebinned CoLoRe cls
    rel_error = (cl_kk_colore - cl_kk_lin_rebin) / cl_kk_lin_rebin
    rel_error_err = err_kk / cl_kk_lin_rebin

    # split the relative error plot into two panels for better visibility of low and high multipoles
    l_lim = 100

    # plot the relative error for low multipoles in the main panel
    ax2.errorbar(ells_colore[ells_colore < l_lim], rel_error[ells_colore < l_lim], 
                yerr=rel_error_err[ells_colore < l_lim], capsize=1.0, elinewidth=0.7, fmt='C0.', markersize=5.0)
    ax2.axhline(0, c='k', ls='--', lw=0.7)
    ax2.set_ylim(-1.0, 1.0)
    ax2.set_xscale('log')
    ax2.set_xlabel(r'$L$')
    ax2.set_ylabel(rf'rel. error $L<{l_lim}$')

    # plot the relative error for high multipoles in a separate panel
    ax3 = ax2.twinx()
    ax3.errorbar(ells_colore[ells_colore >= l_lim], rel_error[ells_colore >= l_lim], 
                yerr=rel_error_err[ells_colore >= l_lim], capsize=1.0, elinewidth=0.7, fmt='C1.', markersize=5.0)
    ax3.set_ylim(-0.2, 0.2)
    ax3.set_xscale('log')
    ax3.set_xlabel(r'$L$')
    ax3.set_ylabel(rf'rel. error $L\geq {l_lim}$')

    fig.suptitle('CMB lensing autospectrum: Theory vs. CoLoRe boxes', y=0.93)

    fig.savefig(f'plots/cls_kk_{box_label}.png', bbox_inches='tight')
    fig.savefig(f'plots/cls_kk_{box_label}.pdf', bbox_inches='tight')
    plt.show()
```

```python
def plot_cumulative_chi2(ells_colore, cl_kk_colore, err_kk, cl_kk_lin_rebin, box_label):

    # compute the cumulative chi-square and p-values for each multipole bin
    chi2_per_bin = ((cl_kk_colore - cl_kk_lin_rebin) / err_kk)**2
    cumulative_chi2 = np.nancumsum(chi2_per_bin)
    cumulative_dof = np.arange(1, len(cumulative_chi2) + 1)
    p_values = 1 - chi2.cdf(cumulative_chi2, df=cumulative_dof)

    # Plotting the cumulative chi-square
    fig, ax1 = plt.subplots(figsize=(6, 4), dpi=120)
    ax2 = ax1.twinx()  # Create a second y-axis for p-values

    # Plot your cumulative chi-square
    ax1.plot(ells_colore, cumulative_chi2, marker='.', linestyle='-', color='C0', label=rf'Cumulative $\chi^2$ ({box_label})')
    ax2.plot(ells_colore, p_values, marker='.', linestyle='-', color='C1', label=r'Cumulative $p$-value', alpha=0.3)

    # Plot the expected theoretical baseline (Expected chi^2 = dof)
    ax1.plot(ells_colore, cumulative_dof, linestyle='--', color='black', label=r'Expected $\chi^2$ (Perfect Fit)')

    # Formatting the plot
    ax1.set_xscale('log')
    ax1.set_yscale('linear') # A log scale on the y-axis is highly recommended since the max value is ~8000
    ax1.set_ylim(0, 100)
    ax1.set_xlabel('$L$')
    ax1.set_ylabel(r'Cumulative $\chi^2$')
    ax2.set_ylabel(r'Cumulative $p$-value')
    ax1.set_title(r'Cumulative $\chi^2$ vs. Multipole $L$')
    ax1.legend(loc='upper left')
    ax2.legend(loc='upper right')

    fig.savefig(f'plots/cls_kk_{box_label}_cumulative_chi2.png', bbox_inches='tight')  
    fig.savefig(f'plots/cls_kk_{box_label}_cumulative_chi2.pdf', bbox_inches='tight')  

    plt.show()
```

# fitting $k_p$

```python
def get_limber_cls_kk_fit(
        params_path, 
        z_pk,
        kp=None,
        kmax=100, 
        lmin=2,
        lmax=2500):
    '''
    Compute the angular kk power spectra for fitting using the Limber approximation.
    '''

    # get linear and non-linear matter power spectrum
    cosmo_dict = load_cosmology(params_path)
    z_max = cosmo_dict["z_max"]

    results_lin = get_camb_results(
        cosmo_dict=cosmo_dict,
        z_pk=z_pk,
        kmax=kmax,
        lmax=lmax,
        nonlinear=False)
    
    # comoving distance of the last scattering surface 
    z_star = results_lin.get_derived_params()['zstar']  # z_star doesn't match z_out until third box
    chi_star = results_lin.comoving_radial_distance(z_star)

    # Hubble parameter and matter density for the kernel
    H0 = cosmo_dict["h"] * 100
    om = cosmo_dict["omega_M"]
    cosmo_params = (H0, om)

    # get the matter power spectrum interpolators for linear and non-linear cases
    pk_interp_lin = results_lin.get_matter_power_interpolator(
        nonlinear=False, 
        hubble_units=False, 
        k_hunit=False,  
        var1='delta_tot', 
        var2='delta_tot')
    
    #store the cls from linear and nonlinear limber approximation integration
    # redshift integration up to zmax
    ells = np.arange(lmin, lmax+1)
    cls = np.zeros(len(ells))

    # manually integrate the limber approximation
    for i, ell in enumerate(ells):

        # integral up to redshift z_source, source plane at z_star
        chi_source = chi_star
        z_cut = z_max
        cls[i] = quad(limber, 1e-4, z_cut, args=(ell, kmax, chi_source, cosmo_params, pk_interp_lin, results_lin, kp))[0]
        

    return {
        'ells': ells,
        'cls': cls,
    }
```

```python
def get_fitting_cls(
    kp,
    params_path,
    box_path,
    z_pk,
    lmin_fit,
    lmax_fit,
    kmax=100.0,
    nbins=30):
    '''
    Compute rebinned data and theory Cls for fitting.
    '''

    # load and rebin colore data
    ells_data, cls_data, err_data = get_colore_cls(box_path)

    _, cls_data, err_data = hybrid_bin_cls(
        ells_data,
        cls_data,
        nbins=nbins,
        ell_min=lmin_fit,
        ell_max=lmax_fit,
    )

    # compute theory
    results = get_limber_cls_kk_fit(
        params_path=params_path,
        z_pk=z_pk,
        kp=kp,
        kmax=kmax,
        lmin=2,
        lmax=2500,
    )

    # rebin theory
    _, cls_theory, _ = hybrid_bin_cls(
        results["ells"],
        results["cls"],
        nbins=nbins,
        ell_min=lmin_fit,
        ell_max=lmax_fit,
    )

    return cls_data, cls_theory, err_data
```

```python
def get_chi2(
        kp, 
        params_path, 
        box_path, 
        z_pk, 
        lmin_fit, 
        lmax_fit):
    '''
    Reduced chi-squared for kk fitting.
    '''

    # compute and rebin data and theory
    cls_data, cls_theory, err_data = get_fitting_cls(
        kp=kp,
        params_path=params_path,
        box_path=box_path,
        z_pk=z_pk,
        lmin_fit=lmin_fit,
        lmax_fit=lmax_fit,
    )

    # remove invalid values
    mask = (
        np.isfinite(cls_data)
        & np.isfinite(cls_theory)
        & np.isfinite(err_data)
        & (err_data > 0)
    )

    cls_data = cls_data[mask]
    cls_theory = cls_theory[mask]
    err_data = err_data[mask]

    # get chi2 statistic
    chi2 = np.sum(
        ((cls_data - cls_theory) / err_data)**2
    )

    return chi2 / (len(cls_data) - 1)  # fitting 1 parameter
```

```python
lmin_fit = 2
lmax_fit = 1000

params_path = '../data/third_test_box/cosmo_params.json'
box_path = '../data/third_test_box/cmb_lensing/cl_kk.npz'

result = minimize_scalar(
    get_chi2, 
    bounds=(0.1, 1.0), 
    method='bounded', 
    args=(
        params_path, 
        box_path, 
        z_pk, 
        lmin_fit, 
        lmax_fit))

best_kp = result.x
chi2_best_kp = result.fun
```
