# Notebook to fill the LSST $n(z), b(z), T(z)$ files from $z=0$ to $z=3.8$ 

The original LSST files that Laura sent me, and got from someone, don't span the same redshift range as CoLoRe does. CoLoRe raises errors if the redshift range of the files don't span all the redshift of the simulation. Thus, one has to manually fill with 0 the remaining redshift of all the files used.

However, if the redshift range of the files span further than the simulation redshift then this is not a problem for CoLoRe. Those redshifts will just be ignored.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load the nz, bz and Tz of LSST galaxies
z_nz, nz = np.loadtxt("nz/nz_GAL_LSST_original.txt", unpack=True)
z_bz, bz = np.loadtxt("bz/bz_GAL_LSST_original.txt", unpack=True)
z_Tz, Tz = np.loadtxt("Tz/Tz_GAL_LSST_original.txt", unpack=True)
```

```python
# fill n(z) values elsewhere with 0
dz = z_nz[1] - z_nz[0]

# build the zero-padding regions
# we use dz/2 offsets to avoid floating-point boundary overlaps
z_prefix = np.arange(0, z_nz[0] - dz / 2, dz)
z_suffix = np.arange(z_nz[-1] + dz, 3.8 + dz / 2, dz)

# concatenate the redshifts and the n(z) values directly
z_nz_new = np.concatenate([z_prefix, z_nz, z_suffix])
nz_new = np.concatenate([np.zeros_like(z_prefix), nz, np.zeros_like(z_suffix)])

# save the perfectly preserved distribution
output_filename = "nz/nz_GAL_LSST_extended.txt"
new_data = np.column_stack((z_nz_new, nz_new))
np.savetxt(output_filename, new_data, fmt="%.6f")
```

```python
# fill b(z) values elsewhere with 0
dz = z_bz[1] - z_bz[0]

# build the zero-padding regions
# we use dz/2 offsets to avoid floating-point boundary overlaps
z_prefix = np.arange(0, z_bz[0] - dz / 2, dz)
z_suffix = np.arange(z_bz[-1] + dz, 3.8 + dz / 2, dz)

# concatenate the redshifts and the n(z) values directly
z_bz_new = np.concatenate([z_prefix, z_bz, z_suffix])
bz_new = np.concatenate([np.zeros_like(z_prefix), bz, np.zeros_like(z_suffix)])

# save the perfectly preserved distribution
output_filename = "bz/bz_GAL_LSST_extended.txt"
new_data = np.column_stack((z_bz_new, bz_new))
np.savetxt(output_filename, new_data, fmt="%.6f")
```

```python
# fill T(z) values elsewhere with 0
dz = z_Tz[1] - z_Tz[0]

# build the zero-padding regions
# we use dz/2 offsets to avoid floating-point boundary overlaps
z_prefix = np.arange(0, z_Tz[0] - dz / 2, dz)
z_suffix = np.arange(z_Tz[-1] + dz, 3.8 + dz / 2, dz)

# concatenate the redshifts and the n(z) values directly
z_Tz_new = np.concatenate([z_prefix, z_Tz, z_suffix])
Tz_new = np.concatenate([np.zeros_like(z_prefix), Tz, np.zeros_like(z_suffix)])

# save the perfectly preserved distribution
output_filename = "Tz/Tz_GAL_LSST_extended.txt"
new_data = np.column_stack((z_Tz_new, Tz_new))
np.savetxt(output_filename, new_data, fmt="%.6f")
```

```python
fig, axs = plt.subplots(1, 3, figsize=(18, 5), sharex=True)

# =====================================================================
# SUBPLOT 1: Redshift Distribution n(z)
# =====================================================================
axs[0].plot(
    z_nz_new,
    nz_new,
    label=rf"Extended (${z_nz_new.min():.2f} \leq z \leq {z_nz_new.max():.2f}$)",
    color="tab:blue",
    lw=2,
    zorder=1,
)
axs[0].fill_between(z_nz_new, nz_new, color="tab:blue", alpha=0.15)
axs[0].scatter(
    z_nz,
    nz,
    label=rf"Original (${z_nz.min():.2f} \leq z \leq {z_nz.max():.2f}$)",
    color="tab:orange",
    edgecolor="black",
    s=25,
    zorder=2,
)
axs[0].set_title("Redshift Distribution Comparison", fontsize=12, pad=10)
axs[0].set_ylabel("$n(z)$", fontsize=12)
axs[0].legend(fontsize=9, loc="upper right")

# =====================================================================
# SUBPLOT 2: Bias Distribution b(z)
# =====================================================================
axs[1].plot(
    z_bz_new,
    bz_new,
    label=rf"Extended (${z_bz_new.min():.2f} \leq z \leq {z_bz_new.max():.2f}$)",
    color="tab:blue",
    lw=2,
    zorder=1,
)
axs[1].fill_between(z_bz_new, bz_new, color="tab:blue", alpha=0.15)
axs[1].scatter(
    z_bz,
    bz,
    label=rf"Original (${z_bz.min():.2f} \leq z \leq {z_bz.max():.2f}$)",
    color="tab:orange",
    edgecolor="black",
    s=25,
    zorder=2,
)
axs[1].set_title("Bias Distribution Comparison", fontsize=12, pad=10)
axs[1].set_ylabel("$b(z)$", fontsize=12)
axs[1].legend(fontsize=9, loc="upper right")

# =====================================================================
# SUBPLOT 3: Threshold Distribution T(z)
# =====================================================================
axs[2].plot(
    z_Tz_new,
    Tz_new,
    label=rf"Extended (${z_Tz_new.min():.2f} \leq z \leq {z_Tz_new.max():.2f}$)",
    color="tab:blue",
    lw=2,
    zorder=1,
)
axs[2].fill_between(z_Tz_new, Tz_new, color="tab:blue", alpha=0.15)
axs[2].scatter(
    z_Tz,
    Tz,
    label=rf"Original (${z_Tz.min():.2f} \leq z \leq {z_Tz.max():.2f}$)",
    color="tab:orange",
    edgecolor="black",
    s=25,
    zorder=2,
)
axs[2].set_title("Threshold Distribution Comparison", fontsize=12, pad=10)
axs[2].set_ylabel("$T(z)$", fontsize=12)
axs[2].legend(fontsize=9, loc="upper right")


# =====================================================================
# GLOBAL STYLING & CLEANUP
# =====================================================================
for ax in axs:
    ax.set_xlabel("Redshift ($z$)", fontsize=11)
    ax.set_xlim(-0.1, 4.0)
    ax.grid(True, linestyle="--", alpha=0.5)

# Tight layout handles padding automatically so text doesn't overlap
plt.tight_layout()
#plt.savefig("combined_distributions.png", dpi=300, bbox_inches="tight")
plt.show()
```
